## 📖 How it works

A résumé lists its languages the way its author wrote them: *deutsch*, *anglais*, *english language*, sometimes with a level (*Muttersprache*, *courant*, *B2*, *TOEIC 905*), often without. Two tagging agents turn that list into one clean tag per language.

| Step | Who | Reads | Answers |
|---|---|---|---|
| **1 · Parse** | HrFlow Parsing | the résumé | `profile["languages"]`: each language as written |
| **2 · Normalize** | Language normalizer, on Programmatic Small | one language name | its ISO 639-1 code: `de` |
| **3 · Rate** | Language level, on Programmatic Large | the language, then the profile: résumé text, experiences, tasks, educations, certifications | a CEFR level, A1 to C2, or *Other* when the profile shows none |
| **4 · Tag** | the pipeline | the two answers | `{"name": "Language", "value": "de:C1"}`, or `de:other` |

The level agent follows written rules, in any language: an explicit CEFR level wins; certificates count by their official tables; a self-assessment counts by its meaning (*native* C2, *fluent* C1, *very good* B2, *good* B1, *school* A2, *basic* A1); a level in progress counts as the level held today. Without a stated level, it estimates from the résumé's language and from where the candidate studied and worked, never from nationality, origin or name.

```python
#@title Engine: the code behind this notebook { display-mode: "form" }
#@markdown The text the agents read, the HrFlow.ai API client, the pipeline and the status cards. Folded: open it with **Show code**.
import html as _html
import json
import time
from getpass import getpass

import requests
from IPython.display import HTML, JSON, display

CONFIG = json.loads(r"""{"product": "Language Tagger", "object": "profile", "words": {"object": "profile", "objects": "profiles", "container": "source", "Container": "Source", "pane": "Résumé", "raw": "Parsed text", "no_raw": "The parser stored no text for this profile.", "empty": "This source holds no profile yet.", "add": "Parse a few résumés into this source, then run the cell again.", "search": "Search a profile", "this": "This profile", "parsed": "HrFlow Parsing", "where": "Connections → My Sources → the source → Source Key", "units": "Languages, as parsed"}, "pipeline": "normalize_then_rate", "steps": ["ISO 639-1", "CEFR level"], "tag_name": "Language", "tag_value": "{code}:{value}", "scale": ["A1", "A2", "B1", "B2", "C1", "C2"]}""")   # the product, what is tagged (profile or job), the steps, the tags' format
OBJECT = CONFIG["object"]
W = CONFIG["words"]                         # the words of this kind of object: profile / source, job / board

# ── 1. The text the agents read (recipe_text.py) ───────────────────────────────────────────────────────────────────
import html
import re

LABELS = {"unit": "Language", "parsed": "Value as parsed", "text": "Resume text", "experiences": "Experiences",
          "tasks": "Tasks", "educations": "Educations", "certifications": "Certifications", "ongoing": "present"}

SECTIONS = [                       # in this order; when the text is too long, the last sections give way first
    {"field": "unit"},
    {"field": "text"},
    {"field": "experiences", "item": "{title} — {company} — {location} — {years}", "detail": "description",
     "recent_first": True, "limit": 10},
    {"field": "tasks", "item": "{name}", "join": "; ", "limit": 40},
    {"field": "educations", "item": "{title} — {school} — {location} — {years}", "detail": "description",
     "recent_first": True, "limit": 8},
    {"field": "certifications", "item": "{name}{value?: }", "join": "; "},
]

MAX_CHARS = 8192      # the characters the tagger reads
DETAIL_CHARS = 600    # the characters kept of each description

MOJIBAKE = re.compile("Ã[\u0080-¿ŒœŠšŸŽžƒˆ˜–—"
                      "‘-„†-•…‰‹›€™]|â€|Â[ -¿]")
CP1252_RUN = re.compile("[À-ÿŒœŠšŸŽžƒˆ˜"
                        "–—‘-„†-•…‰‹›€™\u0080-¿]{2,}")
ZERO_WIDTH = re.compile("[​-‍⁠﻿­]")
HTML_TAG = re.compile(r"<\s*/?\s*[a-zA-Z][a-zA-Z0-9]*\b[^>]*>")
HTML_ENTITY = re.compile(r"&(#\d+|#x[0-9a-fA-F]+|[a-zA-Z]+);")
PLACEHOLDERS = {"", "null", "none", "nan", "undefined", "n/a", "na", "n.a.", "-", "--", "---", ".", "..", "...", "?",
                "x", "xx", "xxx", "()", "/", "//", "#n/a", "#ref!", "#value!", "\\n"}
DATE = re.compile(r"^(\d{4})-(\d{2})-(\d{2})")


def fix_mojibake(s):
    """UTF-8 read as Latin-1 or Windows-1252 (an encoding error, in any language): repaired run by run."""
    if not MOJIBAKE.search(s or ""):
        return s
    for enc in ("latin-1", "cp1252"):
        try:
            return s.encode(enc).decode("utf-8")
        except (UnicodeEncodeError, UnicodeDecodeError):
            pass

    def one(m):
        for enc in ("cp1252", "latin-1"):
            try:
                return m.group(0).encode(enc).decode("utf-8")
            except (UnicodeEncodeError, UnicodeDecodeError):
                continue
        return m.group(0)
    return CP1252_RUN.sub(one, s)


def clean(value, limit=None):
    """A field's text: encoding repaired, HTML removed, empty, placeholder and bullet-only lines dropped."""
    if value is None or isinstance(value, (dict, list)):
        return None
    s = ZERO_WIDTH.sub("", fix_mojibake(str(value))).replace("\\n", "\n")
    if HTML_TAG.search(s) or HTML_ENTITY.search(s):
        s = re.sub(r"<\s*(br|/p|/li|/div)\s*/?>", "\n", s, flags=re.I)
        s = html.unescape(HTML_TAG.sub(" ", s))
    lines = []
    for line in s.splitlines():
        line = " ".join(line.replace(" ", " ").split())
        if line and line.lower() not in PLACEHOLDERS and not re.fullmatch(r"[-•·*–—_=~|]+", line):
            lines.append(line)
    out = "\n".join(lines)
    if limit and len(out) > limit:
        out = out[:limit].rsplit(" ", 1)[0] + " …"
    return out or None


def clean_line(value):
    s = clean(value)
    return " ".join(s.split("\n")) if s else None


def year(value):
    m = DATE.match(value) if isinstance(value, str) else None
    return m.group(1) if m and "1950" <= m.group(1) <= "2027" else None


def years(el, labels):
    a, b = year(el.get("date_start")), year(el.get("date_end"))
    if a and not el.get("date_end"):
        return f"{a}–{labels['ongoing']}"
    return f"{a}–{b}" if a and b and a != b else (a or b)


def place(loc):
    if isinstance(loc, str):
        return clean_line(loc)
    loc = loc or {}
    f = loc.get("fields") or {}
    return clean_line(loc.get("text")) or ", ".join(
        v for v in (clean_line(loc.get("city") or f.get("city")), clean_line(loc.get("country") or f.get("country"))) if v) or None


def fill(template, el, labels):
    """One element laid out by its template: {key} its value; {key?: } the value after ': ' when present; a missing
    value and the separator before it disappear."""
    out = template
    for m in re.finditer(r"\{(\w+)(\?[^}]*)?\}", template):
        key, opt = m.group(1), m.group(2)
        if key == "location":
            v = place(el.get("location"))
        elif key == "years":
            v = years(el, labels)
        else:
            raw = el.get(key)
            v = clean_line(raw) if isinstance(raw, str) else (str(raw) if isinstance(raw, (int, float)) else None)
        out = out.replace(m.group(0), "" if not v else (f"{opt[1:]}{v}" if opt else v), 1)
    out = re.sub(r"(\s*—\s*)+$", "", re.sub(r"^(\s*—\s*)+", "", out))
    return re.sub(r"(\s*—\s*){2,}", " — ", out).strip()


def section(profile, unit, sec, labels):
    """The text of one section, or None when the profile has nothing for it."""
    field, label = sec["field"], labels[sec["field"]]
    if field == "unit":
        if not unit:
            return None
        vals = [f"{k} = {clean_line(v)}" for k, v in (unit.get("fields") or {}).items() if k != "name" and clean_line(v)]
        return f"{label}: {unit.get('name') or '—'}" + (f"\n{labels['parsed']}: " + "; ".join(vals) if vals else "")
    val = profile.get(field)
    if isinstance(val, list):
        els = [e for e in val if isinstance(e, dict)]
        if sec.get("recent_first"):
            els = sorted(els, key=lambda e: (e.get("date_start") or "") if year(e.get("date_start")) else "", reverse=True)
        rows = []
        for el in els[: sec.get("limit") or len(els)]:
            head = fill(sec["item"], el, labels)
            detail = clean(el.get(sec["detail"]), DETAIL_CHARS) if sec.get("detail") else None
            if head or detail:
                rows.append((head, detail))
        if not rows:
            return None
        if sec.get("join"):
            return f"{label}: " + sec["join"].join(h for h, _ in rows if h)
        lines = [f"{label}:"]
        for h, d in rows:
            lines.append(f"- {h}" if h else "-")
            if d:
                lines.append("  " + d.replace("\n", "\n  "))
        return "\n".join(lines)
    t = clean(val)
    if not t:
        return None
    return f"{label}: {t}" if len(t) < 120 and "\n" not in t else f"{label}:\n{t}"


def transform(profile, unit=None, labels=None):
    """The text for one parsed language of a profile, or None when there is nothing to read (no call, no tag)."""
    labels = {**LABELS, **(labels or {})}
    parts = [t for t in (section(profile, unit, sec, labels) for sec in SECTIONS) if t]
    while parts and len("\n\n".join(parts)) > MAX_CHARS:     # the last sections give way first
        excess = len("\n\n".join(parts)) - MAX_CHARS
        if len(parts[-1]) - excess >= 200 or len(parts) == 1:
            parts[-1] = parts[-1][: max(0, len(parts[-1]) - excess - 2)].rsplit(" ", 1)[0] + " …"
        else:
            parts.pop()
    text = "\n\n".join(parts).strip()
    return text or None


def language_entries(profile):
    """One unit per language the profile lists, in the parser's order: (its index in profile["languages"], unit)."""
    for i, entry in enumerate(profile.get("languages") or []):
        if not isinstance(entry, dict) or not str(entry.get("name") or "").strip():
            continue
        yield i, {"name": str(entry["name"]).strip(),
                  "fields": {k: v for k, v in entry.items() if isinstance(v, (str, int, float)) and v not in ("", None)}}


# ── 2. The HrFlow.ai API ───────────────────────────────────────────────────────────────────────────────────────────
KINDS = {   # a profile lives in a source, a job in a board: the same calls, other names
    "profile": {"container": "/source", "list": "/storing/profiles", "keys": "source_keys", "full": "return_profile",
                "one": "/profile/indexing", "param": "source_key", "rows": "profiles"},
    "job": {"container": "/board", "list": "/storing/jobs", "keys": "board_keys", "full": "return_job",
            "one": "/job/indexing", "param": "board_key", "rows": "jobs"},
}
K = KINDS[OBJECT]


class HrFlow:
    """The few public API calls the notebook needs. The only write is creating a taxonomy; nothing writes to a profile
    or a job."""
    API = "https://api.hrflow.ai/v1"

    def __init__(self, api_key, user_email):
        self.s = requests.Session()
        self.s.headers.update({"X-API-KEY": api_key, "X-USER-EMAIL": user_email})

    def _call(self, method, path, **kw):
        for attempt in range(3):
            r = self.s.request(method, f"{self.API}{path}", timeout=120, **kw)
            if r.status_code < 500:
                break
            time.sleep(1.5 * (attempt + 1))
        body = r.json() if r.headers.get("content-type", "").startswith("application/json") else {}
        if r.status_code >= 400 or int(body.get("code") or 200) >= 400:
            raise RuntimeError(f"HrFlow.ai API {r.status_code}: {body.get('message') or r.text[:200]}")
        return body.get("data")

    def container(self, key):
        """The source (profiles) or the board (jobs)."""
        return self._call("GET", K["container"], params={"key": key})

    def objects(self, key, limit=60):
        """The latest profiles of a source, or jobs of a board, from storage: listed even when its search is off."""
        data = self._call("GET", K["list"], params={
            K["keys"]: json.dumps([key]), "limit": limit, "page": 1, K["full"]: "true",
            "sort_by": "created_at", "order_by": "desc"})
        return data if isinstance(data, list) else (data or {}).get(K["rows"]) or []

    def object(self, key, obj_key):
        return self._call("GET", K["one"], params={K["param"]: key, "key": obj_key})

    def taxonomies(self):
        """The workspace's taxonomies, by reference. The API answers the whole list in one call (it has no paging)."""
        rows = self._call("GET", "/taxonomies") or []
        rows = rows.get("taxonomies") or [] if isinstance(rows, dict) else rows
        return {t["reference"]: t for t in rows if t.get("reference")}

    def taxonomy(self, key):
        """One taxonomy, with the agent that uses it ("agent": key and name), if any."""
        return self._call("GET", "/taxonomy", params={"key": key}) or {}

    def create_taxonomy(self, payload):
        return self._call("POST", "/taxonomy", json=payload) or {}

    def tag(self, algorithm_key, texts):
        """One answer per text: (value, label), or (None, None) for Other or for a text that is None (nothing to read)."""
        out, todo = [(None, None)] * len(texts), [i for i, t in enumerate(texts) if t]
        for start in range(0, len(todo), 16):
            part = todo[start:start + 16]
            data = self._call("POST", "/text/tagging", json={
                "algorithm_key": algorithm_key, "texts": [texts[i] for i in part], "top_n": 1, "output_lang": "en"})
            for i, row in zip(part, data or []):
                value = ((row or {}).get("ids") or [None])[0]
                label = ((row or {}).get("tags") or [None])[0]
                if value and str(value).lower() != "other":
                    out[i] = (value, _short(value, label))
        return out


def _short(value, label):
    """The label without its code: "de - German" → "German"; "C1 - Effective operational proficiency" → the words."""
    label = label or ""
    return label.split(" - ", 1)[1].split(";")[0].strip() if label.startswith(f"{value} - ") else label


# ── 3. What gets a tag, and the pipeline ───────────────────────────────────────────────────────────────────────────
def units_of(obj):
    """What gets a tag on this object: each parsed entry when the recipe tags entries (language_entries), else the
    object itself (one unit, None)."""
    if "language_entries" in globals():
        return list(language_entries(obj))
    return [(None, None)]


def _unit_text(obj, unit):
    return transform(obj, unit) if unit is not None else transform(obj)


def _tag_value(**fields):
    return CONFIG["tag_value"].format(**{k: (v if v is not None else "other") for k, v in fields.items()})


def one_agent(i, obj, state, client, keys):
    """One step: the agent on each unit (or on the object). A tag per unit."""
    units = units_of(obj)
    texts = [_unit_text(obj, u) for _, u in units]
    t0 = time.time()
    answers = client.tag(keys[0], texts)
    state["seconds"] = [round(time.time() - t0, 2)]
    state["rows"] = [{"unit": (u or {}).get("name") or W["this"], "answers": [{"value": v, "label": lab}], "text": t, "note": ""}
                     for (_, u), t, (v, lab) in zip(units, texts, answers)]
    state["tags"] = [{"name": CONFIG["tag_name"], "value": _tag_value(unit=r["unit"], value=r["answers"][0]["value"])}
                     for r in state["rows"] if r["text"]]
    return state


def normalize_then_rate(i, obj, state, client, keys):
    """Two steps. 1: each parsed entry's name → a code (a small-tier normalizer); entries with the same code merge.
    2: each code → a rating read from the object (the rater's text names the code on its first line). A tag per code."""
    if i == 0:
        units = units_of(obj)
        t0 = time.time()
        answers = client.tag(keys[0], [(u or {}).get("name") for _, u in units])
        state["seconds"] = [round(time.time() - t0, 2)]
        codes, rows = {}, []
        for (_, u), (code, name) in zip(units, answers):
            merged = bool(code and code in codes)
            if code and not merged:
                codes[code] = {"code": code, "name": name, "row": len(rows)}
            rows.append({"unit": u["name"], "answers": [{"value": code, "label": codes[code]["name"] if code else None}],
                         "text": None, "note": "merged" if merged else ""})
        state["rows"], state["codes"] = rows, list(codes.values())
        return state
    texts = [transform(obj, {"name": f"{c['name']} ({c['code']})", "fields": {}}) for c in state["codes"]]
    t0 = time.time()
    rated = client.tag(keys[1], texts)
    state["seconds"] = state["seconds"][:1] + [round(time.time() - t0, 2)]
    by_code = {c["code"]: (r, t) for c, r, t in zip(state["codes"], rated, texts)}
    for row in state["rows"]:
        code = row["answers"][0]["value"]
        (value, label), text = by_code.get(code, ((None, None), None))
        row["answers"] = row["answers"][:1] + [{"value": value if code else None, "label": label if code else None}]
        row["text"] = text if code and not row["note"] else None
    state["tags"] = [{"name": CONFIG["tag_name"], "value": _tag_value(code=c["code"], value=by_code[c["code"]][0][0])}
                     for c in state["codes"]]
    return state



STEP = globals()[CONFIG["pipeline"]]        # step(i, obj, state, client, keys) → state, for i in range(len(steps))


def run_pipeline(obj, client, keys):
    """Every step on one object: {"rows": one per unit, with each step's answer, "tags": the object's tags}."""
    state = {}
    for i in range(len(CONFIG["steps"])):
        state = STEP(i, obj, state, client, keys)
    return state


# ── 4. How the demo shows an object ────────────────────────────────────────────────────────────────────────────────
def _when(e):
    a, b = (e.get("date_start") or "")[:4], (e.get("date_end") or "")[:4]
    return f"{a} – {b or 'present'}" if a else (b or "")


def _str(v):
    return v if isinstance(v, str) else ""


def shape(obj):
    """The parts of a parsed profile or job the demo shows, as blocks: a paragraph, a timeline or chips."""
    units = [u["name"] for _, u in units_of(obj) if u]
    names = lambda xs: [x.get("name") for x in xs or [] if isinstance(x, dict) and x.get("name")][:14]  # noqa: E731
    if OBJECT == "profile":
        info = obj.get("info") or {}
        exps = sorted(obj.get("experiences") or [], key=lambda e: e.get("date_start") or "", reverse=True)
        blocks = [{"h": "Summary", "p": _str(info.get("summary"))},
                  {"h": "Experience", "tl": [{"when": _when(e), "head": e.get("title"), "sub": " · ".join(
                      x for x in (e.get("company"), ((e.get("location") or {}).get("text") or "")) if x)} for e in exps[:6]]},
                  {"h": "Education", "tl": [{"when": _when(e), "head": e.get("title"), "sub": e.get("school")}
                                             for e in (obj.get("educations") or [])[:4]]},
                  {"h": W["units"], "chips": units} if "language_entries" in globals() else {},
                  {"h": "Skills", "chips": names(obj.get("skills"))}]
        head = {"name": info.get("full_name") or "—", "title": (exps[0].get("title") if exps else None) or "",
                "meta": " · ".join(x for x in ((info.get("location") or {}).get("text") or "",
                                               f"résumé in {obj['text_language'].upper()}" if obj.get("text_language") else "") if x)}
        raw = obj.get("text") or ""
    else:
        blocks = [{"h": "Summary", "p": _str(obj.get("summary"))}]
        blocks += [{"h": s.get("title") or s.get("name") or "Section", "p": _str(s.get("description"))} for s in obj.get("sections") or []]
        blocks += [{"h": h, "p": _str(obj.get(f))} for f, h in (("responsibilities", "Responsibilities"), ("requirements", "Requirements"),
                                                                 ("culture", "Culture"), ("benefits", "Benefits"))]
        blocks += [{"h": W["units"], "chips": units} if "language_entries" in globals() else {},
                   {"h": "Skills", "chips": names(obj.get("skills"))}]
        head = {"name": obj.get("name") or "—", "title": ((obj.get("location") or {}).get("text") or ""), "meta": ""}
        raw = "\n\n".join(b["p"] for b in blocks if b.get("p"))
    blocks = [b for b in blocks if b and (b.get("p") or b.get("tl") or b.get("chips") or b.get("h") == W["units"])]
    return {"key": obj.get("key"), **head, "blocks": blocks, "raw": raw[:8000], "units": units,
            "taggable": bool(units) or "language_entries" not in globals()}


# ── 5. The notebook's cards and secrets ────────────────────────────────────────────────────────────────────────────
_CARD = """<link rel="stylesheet" href="https://fonts.googleapis.com/css2?family=Lato:wght@400;700;900&display=swap">
<style>html,body{margin:0;background:transparent}.card{font-family:Lato,Arial,sans-serif;border:1px solid #e6e9ec;border-radius:12px;
background:#fff;color:#1b1f23;padding:14px 18px;max-width:760px}.h{font-weight:900;font-size:15px;margin-bottom:8px;color:#1b1f23}
.c{display:grid;grid-template-columns:14px 220px 1fr;align-items:center;gap:8px;padding:6px 0;border-top:1px solid #f0f2f4;font-size:14px}
.c i{width:9px;height:9px;border-radius:50%;background:#12805c}.c.err i{background:#b42318}.c.todo i{background:#d5dbe0}
.c b{color:#1b1f23}.c span{color:#4a5158}.c.err span{color:#b42318}.n{font-size:13px;color:#4a5158;margin-top:10px}</style>
<div class="card"><div class="h">{title}</div>{rows}{note}</div>"""


def _card(title, rows, note=""):
    """A status card. rows: (state, name, text), state in ok | err | todo."""
    body = "".join(f'<div class="c {s}"><i></i><b>{_html.escape(n)}</b><span>{_html.escape(t)}</span></div>' for s, n, t in rows)
    note = f'<div class="n">{_html.escape(note)}</div>' if note else ""
    display(HTML(_CARD.replace("{title}", _html.escape(title)).replace("{rows}", body).replace("{note}", note)))


def _secret(name):
    """A Colab secret, or a hidden prompt outside Colab. Never printed, never stored."""
    try:
        from google.colab import userdata
        value = userdata.get(name)
        if value:
            return value
    except Exception:
        pass
    return getpass(f"{name}: ")


def _safe(fn):
    """A callback for the page: its answer, or the error message the page shows where it happened."""
    def run(*args):
        try:
            return JSON(fn(*args))
        except Exception as e:
            return JSON({"error": str(e)})
    return run


def _register(name, fn):
    try:
        from google.colab import output
        output.register_callback(f"tt.{name}", _safe(fn))
        return True
    except ImportError:
        return False


print(f"Engine ready: the HrFlow.ai client, transform() and run_pipeline() for each {OBJECT}.")
```

## 🧰 2 · Set up your workspace

Once per workspace. The panel shows what your workspace already has and offers only what is missing:

- **Taxonomies.** *ISO 639-1 Languages* and *CEFR Levels*, created by the **Create** button. The notebook checks your workspace again just before, so a taxonomy is never created twice.

```python
#@title Your workspace: taxonomies and agents { display-mode: "form" }
#@markdown Shows what your workspace already has and offers only what is missing. Nothing is created unless you press **Create**.
import json
import threading

TAXONOMIES = json.loads(r"""[{"value": "iso-639-1", "label": "ISO 639-1 Languages", "name": "ISO 639-1 Languages", "reference": "language-tags-iso-639-1", "language": "en", "description": "ISO 639-1: the two-letter code of a language, from the list of the ISO 639-2 Registration Authority (Library of Congress). Input: one language name as a résumé parser wrote it, in any language or spelling.", "tags": [{"value": "aa", "label": "aa - Afar"}, {"value": "ab", "label": "ab - Abkhazian"}, {"value": "af", "label": "af - Afrikaans"}, {"value": "ak", "label": "ak - Akan"}, {"value": "sq", "label": "sq - Albanian"}, {"value": "am", "label": "am - Amharic"}, {"value": "ar", "label": "ar - Arabic"}, {"value": "an", "label": "an - Aragonese"}, {"value": "hy", "label": "hy - Armenian"}, {"value": "as", "label": "as - Assamese"}, {"value": "av", "label": "av - Avaric"}, {"value": "ae", "label": "ae - Avestan"}, {"value": "ay", "label": "ay - Aymara"}, {"value": "az", "label": "az - Azerbaijani"}, {"value": "ba", "label": "ba - Bashkir"}, {"value": "bm", "label": "bm - Bambara"}, {"value": "eu", "label": "eu - Basque"}, {"value": "be", "label": "be - Belarusian"}, {"value": "bn", "label": "bn - Bengali"}, {"value": "bi", "label": "bi - Bislama"}, {"value": "bs", "label": "bs - Bosnian"}, {"value": "br", "label": "br - Breton"}, {"value": "bg", "label": "bg - Bulgarian"}, {"value": "my", "label": "my - Burmese"}, {"value": "ca", "label": "ca - Catalan; Valencian"}, {"value": "ch", "label": "ch - Chamorro"}, {"value": "ce", "label": "ce - Chechen"}, {"value": "zh", "label": "zh - Chinese"}, {"value": "cu", "label": "cu - Church Slavic; Old Slavonic; Church Slavonic; Old Bulgarian; Old Church Slavonic"}, {"value": "cv", "label": "cv - Chuvash"}, {"value": "kw", "label": "kw - Cornish"}, {"value": "co", "label": "co - Corsican"}, {"value": "cr", "label": "cr - Cree"}, {"value": "cs", "label": "cs - Czech"}, {"value": "da", "label": "da - Danish"}, {"value": "dv", "label": "dv - Divehi; Dhivehi; Maldivian"}, {"value": "nl", "label": "nl - Dutch; Flemish"}, {"value": "dz", "label": "dz - Dzongkha"}, {"value": "en", "label": "en - English"}, {"value": "eo", "label": "eo - Esperanto"}, {"value": "et", "label": "et - Estonian"}, {"value": "ee", "label": "ee - Ewe"}, {"value": "fo", "label": "fo - Faroese"}, {"value": "fj", "label": "fj - Fijian"}, {"value": "fi", "label": "fi - Finnish"}, {"value": "fr", "label": "fr - French"}, {"value": "fy", "label": "fy - Western Frisian"}, {"value": "ff", "label": "ff - Fulah"}, {"value": "ka", "label": "ka - Georgian"}, {"value": "de", "label": "de - German"}, {"value": "gd", "label": "gd - Gaelic; Scottish Gaelic"}, {"value": "ga", "label": "ga - Irish"}, {"value": "gl", "label": "gl - Galician"}, {"value": "gv", "label": "gv - Manx"}, {"value": "el", "label": "el - Modern Greek (1453-)"}, {"value": "gn", "label": "gn - Guarani"}, {"value": "gu", "label": "gu - Gujarati"}, {"value": "ht", "label": "ht - Haitian; Haitian Creole"}, {"value": "ha", "label": "ha - Hausa"}, {"value": "he", "label": "he - Hebrew"}, {"value": "hz", "label": "hz - Herero"}, {"value": "hi", "label": "hi - Hindi"}, {"value": "ho", "label": "ho - Hiri Motu"}, {"value": "hr", "label": "hr - Croatian"}, {"value": "hu", "label": "hu - Hungarian"}, {"value": "ig", "label": "ig - Igbo"}, {"value": "is", "label": "is - Icelandic"}, {"value": "io", "label": "io - Ido"}, {"value": "ii", "label": "ii - Sichuan Yi; Nuosu"}, {"value": "iu", "label": "iu - Inuktitut"}, {"value": "ie", "label": "ie - Interlingue; Occidental"}, {"value": "ia", "label": "ia - Interlingua (International Auxiliary Language Association)"}, {"value": "id", "label": "id - Indonesian"}, {"value": "ik", "label": "ik - Inupiaq"}, {"value": "it", "label": "it - Italian"}, {"value": "jv", "label": "jv - Javanese"}, {"value": "ja", "label": "ja - Japanese"}, {"value": "kl", "label": "kl - Kalaallisut; Greenlandic"}, {"value": "kn", "label": "kn - Kannada"}, {"value": "ks", "label": "ks - Kashmiri"}, {"value": "kr", "label": "kr - Kanuri"}, {"value": "kk", "label": "kk - Kazakh"}, {"value": "km", "label": "km - Central Khmer"}, {"value": "ki", "label": "ki - Kikuyu; Gikuyu"}, {"value": "rw", "label": "rw - Kinyarwanda"}, {"value": "ky", "label": "ky - Kirghiz; Kyrgyz"}, {"value": "kv", "label": "kv - Komi"}, {"value": "kg", "label": "kg - Kongo"}, {"value": "ko", "label": "ko - Korean"}, {"value": "kj", "label": "kj - Kuanyama; Kwanyama"}, {"value": "ku", "label": "ku - Kurdish"}, {"value": "lo", "label": "lo - Lao"}, {"value": "la", "label": "la - Latin"}, {"value": "lv", "label": "lv - Latvian"}, {"value": "li", "label": "li - Limburgan; Limburger; Limburgish"}, {"value": "ln", "label": "ln - Lingala"}, {"value": "lt", "label": "lt - Lithuanian"}, {"value": "lb", "label": "lb - Luxembourgish; Letzeburgesch"}, {"value": "lu", "label": "lu - Luba-Katanga"}, {"value": "lg", "label": "lg - Ganda"}, {"value": "mk", "label": "mk - Macedonian"}, {"value": "mh", "label": "mh - Marshallese"}, {"value": "ml", "label": "ml - Malayalam"}, {"value": "mi", "label": "mi - Maori"}, {"value": "mr", "label": "mr - Marathi"}, {"value": "ms", "label": "ms - Malay"}, {"value": "mg", "label": "mg - Malagasy"}, {"value": "mt", "label": "mt - Maltese"}, {"value": "mn", "label": "mn - Mongolian"}, {"value": "na", "label": "na - Nauru"}, {"value": "nv", "label": "nv - Navajo; Navaho"}, {"value": "nr", "label": "nr - South Ndebele"}, {"value": "nd", "label": "nd - North Ndebele"}, {"value": "ng", "label": "ng - Ndonga"}, {"value": "ne", "label": "ne - Nepali"}, {"value": "nn", "label": "nn - Norwegian Nynorsk"}, {"value": "nb", "label": "nb - Norwegian Bokmål"}, {"value": "no", "label": "no - Norwegian"}, {"value": "ny", "label": "ny - Chichewa; Chewa; Nyanja"}, {"value": "oc", "label": "oc - Occitan (post 1500)"}, {"value": "oj", "label": "oj - Ojibwa"}, {"value": "or", "label": "or - Oriya"}, {"value": "om", "label": "om - Oromo"}, {"value": "os", "label": "os - Ossetian; Ossetic"}, {"value": "pa", "label": "pa - Panjabi; Punjabi"}, {"value": "fa", "label": "fa - Persian"}, {"value": "pi", "label": "pi - Pali"}, {"value": "pl", "label": "pl - Polish"}, {"value": "pt", "label": "pt - Portuguese"}, {"value": "ps", "label": "ps - Pushto; Pashto"}, {"value": "qu", "label": "qu - Quechua"}, {"value": "rm", "label": "rm - Romansh"}, {"value": "ro", "label": "ro - Romanian; Moldavian; Moldovan"}, {"value": "rn", "label": "rn - Rundi"}, {"value": "ru", "label": "ru - Russian"}, {"value": "sg", "label": "sg - Sango"}, {"value": "sa", "label": "sa - Sanskrit"}, {"value": "si", "label": "si - Sinhala; Sinhalese"}, {"value": "sk", "label": "sk - Slovak"}, {"value": "sl", "label": "sl - Slovenian"}, {"value": "se", "label": "se - Northern Sami"}, {"value": "sm", "label": "sm - Samoan"}, {"value": "sn", "label": "sn - Shona"}, {"value": "sd", "label": "sd - Sindhi"}, {"value": "so", "label": "so - Somali"}, {"value": "st", "label": "st - Sotho, Southern"}, {"value": "es", "label": "es - Spanish; Castilian"}, {"value": "sc", "label": "sc - Sardinian"}, {"value": "sr", "label": "sr - Serbian"}, {"value": "ss", "label": "ss - Swati"}, {"value": "su", "label": "su - Sundanese"}, {"value": "sw", "label": "sw - Swahili"}, {"value": "sv", "label": "sv - Swedish"}, {"value": "ty", "label": "ty - Tahitian"}, {"value": "ta", "label": "ta - Tamil"}, {"value": "tt", "label": "tt - Tatar"}, {"value": "te", "label": "te - Telugu"}, {"value": "tg", "label": "tg - Tajik"}, {"value": "tl", "label": "tl - Tagalog"}, {"value": "th", "label": "th - Thai"}, {"value": "bo", "label": "bo - Tibetan"}, {"value": "ti", "label": "ti - Tigrinya"}, {"value": "to", "label": "to - Tonga (Tonga Islands)"}, {"value": "tn", "label": "tn - Tswana"}, {"value": "ts", "label": "ts - Tsonga"}, {"value": "tk", "label": "tk - Turkmen"}, {"value": "tr", "label": "tr - Turkish"}, {"value": "tw", "label": "tw - Twi"}, {"value": "ug", "label": "ug - Uighur; Uyghur"}, {"value": "uk", "label": "uk - Ukrainian"}, {"value": "ur", "label": "ur - Urdu"}, {"value": "uz", "label": "uz - Uzbek"}, {"value": "ve", "label": "ve - Venda"}, {"value": "vi", "label": "vi - Vietnamese"}, {"value": "vo", "label": "vo - Volapük"}, {"value": "cy", "label": "cy - Welsh"}, {"value": "wa", "label": "wa - Walloon"}, {"value": "wo", "label": "wo - Wolof"}, {"value": "xh", "label": "xh - Xhosa"}, {"value": "yi", "label": "yi - Yiddish"}, {"value": "yo", "label": "yo - Yoruba"}, {"value": "za", "label": "za - Zhuang; Chuang"}, {"value": "zu", "label": "zu - Zulu"}]}, {"value": "cefr-levels", "label": "CEFR Levels", "name": "CEFR Levels", "reference": "language-tags-cefr-levels", "language": "en", "description": "CEFR levels of one language: A1 and A2 (basic user), B1 and B2 (independent user), C1 and C2 (proficient user). Answer exactly one of the six levels for the language named on the first line, or Other when the résumé gives neither a level nor a clue for it.", "tags": [{"value": "A1", "label": "A1 - Breakthrough", "description": "Can understand and use familiar everyday expressions and very basic phrases aimed at the satisfaction of needs of a concrete type. Can introduce themselves to others and can ask and answer questions about personal details such as where they live, people they know and things they have. Can interact in a simple way provided the other person talks slowly and clearly and is prepared to help. On a résumé, in any language: an explicit A1, or a basic or beginner knowledge of the language."}, {"value": "A2", "label": "A2 - Waystage", "description": "Can understand sentences and frequently used expressions related to areas of most immediate relevance (e.g. very basic personal and family information, shopping, local geography, employment). Can communicate in simple and routine tasks requiring a simple and direct exchange of information on familiar and routine matters. Can describe in simple terms aspects of their background, immediate environment and matters in areas of immediate need. On a résumé, in any language: an explicit A2, or a school or elementary level; TOEIC 225-549."}, {"value": "B1", "label": "B1 - Threshold", "description": "Can understand the main points of clear standard input on familiar matters regularly encountered in work, school, leisure, etc. Can deal with most situations likely to arise while travelling in an area where the language is spoken. Can produce simple connected text on topics that are familiar or of personal interest. Can describe experiences and events, dreams, hopes and ambitions and briefly give reasons and explanations for opinions and plans. On a résumé, in any language: an explicit B1, or a good, intermediate or conversational command; TOEIC 550-784, IELTS 4.0-5.0, TOEFL iBT 42-71, or a B1 certificate."}, {"value": "B2", "label": "B2 - Vantage", "description": "Can understand the main ideas of complex text on both concrete and abstract topics, including technical discussions in their field of specialisation. Can interact with a degree of fluency and spontaneity that makes regular interaction with native speakers quite possible without strain for either party. Can produce clear, detailed text on a wide range of subjects and explain a viewpoint on a topical issue giving the advantages and disadvantages of various options. On a résumé, in any language: an explicit B2, or a very good or professional command; TOEIC 785-944, IELTS 5.5-6.5, TOEFL iBT 72-94, Cambridge First, or a B2 certificate."}, {"value": "C1", "label": "C1 - Effective operational proficiency", "description": "Can understand a wide range of demanding, longer texts, and recognise implicit meaning. Can express ideas fluently and spontaneously without much obvious searching for expressions. Can use language flexibly and effectively for social, academic and professional purposes. Can produce clear, well-structured, detailed text on complex subjects, showing controlled use of organisational patterns, connectors and cohesive devices. On a résumé, in any language: an explicit C1, or fluent, business fluent or a working language; TOEIC 945+, IELTS 7.0+, TOEFL iBT 95+, Cambridge Advanced, or a C1 certificate."}, {"value": "C2", "label": "C2 - Mastery", "description": "Can understand with ease virtually everything heard or read. Can summarise information from different spoken and written sources, reconstructing arguments and accounts in a coherent presentation. Can express themselves spontaneously, very fluently and precisely, differentiating finer shades of meaning even in the most complex situations. On a résumé, in any language: an explicit C2, or native, mother tongue or bilingual; Cambridge Proficiency, or a C2 certificate."}]}]""")        # what Create sends, in this order
AGENT_SPECS = json.loads(r"""[{"role": "iso", "title": "Language normalizer", "name": "Language normalizer (ISO 639-1)", "card": "Programmatic Small", "description": "Normalizes one language name, as a résumé parser wrote it, to its ISO 639-1 code.", "rows": 2}, {"role": "cefr", "title": "Language level", "name": "Language level (CEFR)", "card": "Programmatic Large", "description": "Input: one language a candidate lists, named on the first line exactly as the résumé parser wrote it (any language or spelling, possibly misspelled or in the résumé's own language), then the candidate's profile, in any language.\nTask: the candidate's CEFR level in that language only. Work out which language the first line names, then read what the profile says about it, whatever language it is written in. Ignore every other language.\nDecide in this order:\n1. An explicit CEFR level for this language wins (\"B2+\" is B2). A code that is not a CEFR level (a school track, a class number) is not a level.\n2. A certificate, by its official table: TOEIC listening and reading 945+ C1, 785+ B2, 550+ B1, 225+ A2; IELTS 7.0+ C1, 5.5-6.5 B2, 4.0-5.0 B1; TOEFL iBT 95+ C1, 72+ B2, 42+ B1; Cambridge First B2, Advanced C1, Proficiency C2; others: the level they certify.\n3. A self-assessment, in any language, by its meaning: native, mother tongue, bilingual: C2; fluent, business fluent, working language: C1; very good, professional: B2; good, intermediate, conversational: B1; school level, elementary: A2; basic, beginner: A1.\n4. A level in progress or awaited counts as the last level completed; a level only being trained for, one level below it.\n5. One level phrase shared by several languages applies to each.\n6. No level stated: estimate it from the résumé's language, where and how long the candidate studied and worked, and how they use the language at work; never from nationality, origin or name. C2: the résumé is written in this language and the candidate studied or spent most of the career where it is spoken. C1: only one of these: the résumé is written in it; a degree taken in it or there; 2 years or more of work there. B2: used at work (clients, meetings, an international team) with less time there.\n7. Listed with no level and no such clue, or not a language the candidate speaks (a nationality, a company's country): Other.", "rows": 12}]""")      # the agent of each taxonomy, as you create it in AI Studio
SETUP_HTML = r"""<link rel="stylesheet" href="https://fonts.googleapis.com/css2?family=Lato:wght@400;700;900&family=Noto+Sans+Mono:wght@400;600&display=swap">
<style>
  html, body { margin:0; background:transparent }
  .su { --brand:#009db9; --brand-d:#00748a; --tint:#eaf6f8; --ink:#1b1f23; --ink2:#4a5158; --muted:#6b7280; --rule:#e6e9ec; --soft:#f4f6f8;
        --ok:#12805c; --ok-bg:#e3f5ec; --err:#b42318; --err-bg:#fde8e7;
        font-family:"Lato","Helvetica Neue",Arial,sans-serif; color:var(--ink); max-width:900px; border:1px solid var(--rule);
        border-radius:14px; background:#fff; overflow:hidden }
  .su .top { display:flex; align-items:center; gap:10px; padding:14px 20px; border-bottom:1px solid var(--rule); background:var(--soft) }
  .su .top b { font-size:16px; font-weight:900; color:var(--ink) }
  .su .pill { margin-left:auto; font-size:12px; font-weight:700; border-radius:999px; padding:3px 10px; color:var(--muted); background:#fff; border:1px solid var(--rule) }
  .su .pill.ok { color:var(--ok); background:var(--ok-bg); border-color:transparent }
  .su .sec { padding:16px 20px; border-top:1px solid var(--rule) }
  .su .sec:first-of-type { border-top:0 }
  .su .k { font-size:11.5px; font-weight:700; letter-spacing:.1em; text-transform:uppercase; color:var(--brand-d); margin-bottom:8px }
  .su .line { display:grid; grid-template-columns:14px 200px 1fr; gap:10px; align-items:center; padding:7px 0; font-size:14px; border-top:1px solid #f0f2f4 }
  .su .line:first-of-type { border-top:0 }
  .su i.d { width:9px; height:9px; border-radius:50%; background:#d5dbe0; display:block }
  .su i.d.ok { background:var(--ok) } .su i.d.err { background:var(--err) }
  .su .line b { font-weight:700 } .su .line span { color:var(--ink2) }
  .su .act { display:flex; align-items:center; gap:12px; margin-top:12px; flex-wrap:wrap }
  .su .act small { color:var(--muted); font-size:12.5px }
  .su button { font-family:inherit; font-size:13.5px; font-weight:700; border-radius:9px; padding:8px 16px; cursor:pointer;
               color:#fff; background:var(--brand); border:1px solid var(--brand) }
  .su button:hover { background:var(--brand-d); border-color:var(--brand-d) }
  .su button.ghost { color:var(--brand-d); background:#fff; border-color:#9fd6e1 }
  .su button.ghost:hover { background:var(--tint) }
  .su button.copy { font-size:12.5px; padding:5px 12px; min-width:74px }
  .su button.copy.done { color:var(--ok); border-color:#9ad6bd }
  .su button[disabled] { opacity:.55; cursor:default }
  .su .agent { border:1px solid var(--rule); border-radius:12px; margin-top:10px; overflow:hidden }
  .su .ah { display:flex; align-items:center; gap:12px; padding:11px 16px; background:#fff }
  .su .num { width:24px; height:24px; border-radius:50%; background:var(--brand); color:#fff; font-weight:900; font-size:13px;
             display:flex; align-items:center; justify-content:center; flex:none }
  .su .agent.found .num { background:var(--ok) }
  .su .agent.wait .num { background:#c4cbd2 }
  .su .ah b { font-size:15px; font-weight:900 } .su .ah em { font-style:normal; color:var(--muted); font-size:13px }
  .su .ah .st { margin-left:auto; font-size:13px; color:var(--ink2); text-align:right }
  .su .agent.found .ah .st { color:var(--ok); font-weight:700 }
  .su .steps { border-top:1px solid var(--rule); background:#fbfcfd; padding:6px 16px 12px }
  .su .step { display:grid; grid-template-columns:22px 110px 1fr auto; gap:10px; align-items:start; padding:8px 0; font-size:14px; border-top:1px solid #eef1f3 }
  .su .step:first-child { border-top:0 }
  .su .step .n { color:var(--brand-d); font-weight:900 }
  .su .step .l { color:var(--muted); font-size:12px; font-weight:700; text-transform:uppercase; letter-spacing:.06em; padding-top:2px }
  .su textarea { width:100%; box-sizing:border-box; font-family:"Noto Sans Mono",ui-monospace,Menlo,monospace; font-size:12px; line-height:1.5;
                 color:var(--ink); background:#fff; border:1px solid var(--rule); border-radius:8px; padding:8px 10px; resize:vertical }
  .su .warn { margin-top:8px; font-size:13px; color:var(--ink2); background:var(--tint); border-radius:8px; padding:8px 12px }
  .su .msg { font-size:13.5px; margin-top:10px } .su .msg.err { color:var(--err) } .su .msg.ok { color:var(--ok) }
  .su .foot { padding:12px 20px; background:var(--ok-bg); color:var(--ok); font-weight:700; font-size:14px; border-top:1px solid var(--rule) }
</style>
<div class="su" id="su" role="region" aria-label="Language Tagger setup"></div>
<script>
(function () {
  'use strict';
  const S = {{STATE}};                               // inlined by the cell: Colab drops JSON script blocks from outputs
  const root = document.getElementById('su');
  const esc = (s) => String(s == null ? '' : s).replace(/[&<>"']/g, (c) => ({ '&': '&amp;', '<': '&lt;', '>': '&gt;', '"': '&quot;', "'": '&#39;' }[c]));
  let busy = '', msg = null;

  async function call(name) {
    const r = await google.colab.kernel.invokeFunction('tt.setup.' + name, [], {});
    const d = r && r.data && r.data['application/json'];
    if (!d) throw new Error('No answer from the notebook: run the cells above, then this one again.');
    if (d.error) throw new Error(d.error);
    return d;
  }

  async function act(name) {
    if (busy) return;
    busy = name; msg = null; render();
    try {
      const d = await call(name);
      S.taxonomies = d.taxonomies;
      if (name === 'create') msg = ['ok', d.created.length ? 'Created: ' + d.created.join(', ') + '.' : 'Nothing to create: ' + (S.taxonomies.length === 1 ? 'it was' : 'every taxonomy was') + ' already there.'];
    } catch (e) { msg = ['err', e.message]; }
    busy = ''; render();
  }

  function agentHtml(t, spec, i) {
    const state = t.agent ? 'found' : (t.key ? 'todo' : 'wait');
    const st = t.agent ? 'Found · ' + esc(t.agent.name) : (t.key ? 'To create in AI Studio' : 'Waiting for its taxonomy');
    let h = '<div class="agent ' + state + '"><div class="ah"><div class="num">' + (i + 1) + '</div><div><b>' + esc(spec.title) +
            '</b> <em>· ' + esc(spec.card) + '</em></div><div class="st">' + st + '</div></div>';
    if (state === 'todo') {
      const n = 'n' + i, d = 'd' + i;
      h += '<div class="steps">' +
        '<div class="step"><span class="n">1</span><span class="l">Card</span><span>AI Studio → Algorithms Marketplace → Tagging Algorithms → <b>' + esc(spec.card) + '</b></span><span></span></div>' +
        '<div class="step"><span class="n">2</span><span class="l">Name</span><span id="' + n + '">' + esc(spec.name) + '</span><button class="ghost copy" data-copy="' + n + '">Copy</button></div>' +
        '<div class="step"><span class="n">3</span><span class="l">Taxonomy</span><span><b>' + esc(t.label) + '</b> (type its name in the list)</span><span></span></div>' +
        '<div class="step"><span class="n">4</span><span class="l">Description</span><textarea id="' + d + '" readonly rows="' + spec.rows + '" aria-label="Description of ' + esc(spec.title) + '">' + esc(spec.description) + '</textarea><button class="ghost copy" data-copy="' + d + '">Copy</button></div>' +
        '<div class="step"><span class="n">5</span><span class="l">Then</span><span><b>Create Algorithm</b>, then <b>Deploy</b> on the agent\'s page.</span><span></span></div>' +
        '<div class="warn">Create it once. Its taxonomy then serves this agent only, and this panel finds the agent when you press <b>Check again</b>.</div></div>';
    }
    return h + '</div>';
  }

  function render() {
    if (S.error) {
      root.innerHTML = '<div class="top"><b>Your workspace</b><span class="pill">Not read</span></div><div class="sec"><div class="msg err">' +
        esc(S.error) + '</div><div class="msg">Check the Connect cell, then run this cell again.</div></div>';
      return;
    }
    const tx = S.taxonomies || [], missing = tx.filter((t) => !t.key), agentsMissing = tx.filter((t) => !t.agent);
    const off = !S.live ? ' disabled title="Runs in Google Colab"' : '';
    let h = '<div class="top"><b>Your workspace</b><span class="pill' + (tx.length && !missing.length && !agentsMissing.length ? ' ok">Set up' : '">' +
            (missing.length ? 'Taxonomies to create' : 'Agents to create')) + '</span></div>';
    h += '<div class="sec"><div class="k">Taxonomies</div>' + tx.map((t) =>
      '<div class="line"><i class="d' + (t.key ? ' ok' : '') + '"></i><b>' + esc(t.label) + '</b><span>' +
      (t.key ? 'in your workspace · ' + esc(t.size) + ' tags · ' + esc(t.key) : 'not created yet') + '</span></div>').join('');
    if (missing.length) {
      h += '<div class="act"><button data-act="create"' + off + '>' + (busy === 'create' ? 'Creating…' : 'Create ' + (missing.length === 1 ? missing[0].label : missing.length + ' taxonomies')) +
           '</button><small>Creates only what is missing, after checking your workspace again.</small></div>';
    } else if (tx.length) {                            // shown greyed out, so that nobody looks for it
      h += '<div class="act"><button disabled>Create ' + (tx.length === 1 ? tx[0].label : tx.length + ' taxonomies') + '</button><small>Nothing to create: ' +
           (tx.length === 1 ? 'it is' : 'every taxonomy is') + ' already in your workspace.</small></div>';
    }
    h += '</div><div class="sec"><div class="k">Tagging agents · created in AI Studio</div>' +
         tx.map((t, i) => agentHtml(t, S.agents[i], i)).join('');
    if (tx.some((t) => t.key && !t.agent)) {           // an agent to create, its taxonomy already there
      h += '<div class="act"><button class="ghost" data-act="status"' + off + '>' + (busy === 'status' ? 'Checking…' : 'Check again') +
           '</button><small>After creating an agent: the panel then shows it, by name.</small></div>';
    }
    if (msg) h += '<div class="msg ' + msg[0] + '">' + esc(msg[1]) + '</div>';
    h += '</div>';
    if (tx.length && !missing.length && !agentsMissing.length) h += '<div class="foot">Your workspace is set up. Next: 3 · Check the agents.</div>';
    root.innerHTML = h;
  }

  root.addEventListener('click', async (ev) => {
    const b = ev.target.closest('button');
    if (!b || b.disabled) return;
    if (b.dataset.act) return act(b.dataset.act);
    if (b.dataset.copy) {
      const el = document.getElementById(b.dataset.copy), text = el.value !== undefined ? el.value : el.textContent;
      let ok = false;
      try { await navigator.clipboard.writeText(text); ok = true; } catch (e) {}
      if (!ok) {                                   // a frame without clipboard access: the older way
        const t = document.createElement('textarea'); t.value = text; document.body.appendChild(t); t.select();
        try { ok = document.execCommand('copy'); } catch (e) {}
        t.remove();
      }
      b.textContent = ok ? 'Copied' : 'Select, then Ctrl+C'; b.classList.toggle('done', ok);
      setTimeout(() => { b.textContent = 'Copy'; b.classList.remove('done'); }, 1800);
    }
  });
  try { render(); } catch (e) { root.textContent = 'The panel could not be drawn: ' + e.message; }
})();
</script>
"""
AGENTS = {}                                           # role → the agent found: its key and name
_CREATING = threading.Lock()


def _workspace():
    """Each taxonomy of the setup, found by its reference, and the agent that uses it."""
    found = hrflow.taxonomies()
    rows = []
    for t, spec in zip(TAXONOMIES, AGENT_SPECS):
        have = found.get(t["reference"])
        agent = hrflow.taxonomy(have["key"]).get("agent") if have else None
        agent = {"key": agent["key"], "name": agent.get("name") or ""} if agent and agent.get("key") else None
        if agent:
            AGENTS[spec["role"]] = agent
        else:
            AGENTS.pop(spec["role"], None)
        rows.append({"label": t["label"], "key": have and have["key"], "size": have and have.get("size"), "agent": agent})
    return {"taxonomies": rows}


def _create_missing():
    """Creates the taxonomies the workspace lacks, after checking it again; one run at a time."""
    if not _CREATING.acquire(blocking=False):
        return {"error": "The taxonomies are being created: wait a few seconds."}
    try:
        found, created = hrflow.taxonomies(), []
        for t in TAXONOMIES:
            if t["reference"] not in found:
                hrflow.create_taxonomy(t)
                created.append(t["label"])
        return {**_workspace(), "created": created}
    finally:
        _CREATING.release()


if not globals().get("CONNECTED"):
    print("Run the Connect cell first.")
else:
    try:
        _state = _workspace()
    except Exception as e:
        _state = {"error": str(e), "taxonomies": []}
    _live = _register("setup.status", _workspace) and _register("setup.create", _create_missing)
    _state.update(agents=AGENT_SPECS, live=_live)
    display(HTML(SETUP_HTML.replace("{{STATE}}", json.dumps(_state, ensure_ascii=False).replace("</", "<\\/"))))
```

## ✅ 3 · Check the agents

Each agent answers a few questions. Green: the answer expected. Grey: another answer. Red: an error, most often an agent not deployed yet (AI Studio → the agent → **Deploy**).

```python
#@title Check the agents { display-mode: "form" }
#@markdown Asks each agent found in step 2 a few questions.
_CHECKS = json.loads(r"""[{"role": "iso", "title": "Language normalizer", "texts": ["deutsch", "anglais", "english language", "español"], "expected": ["de", "en", "en", "es"], "shown": ["deutsch", "anglais", "english language", "español"]}, {"role": "cefr", "title": "Language level", "texts": ["Language: German (de)\n\nResume text:\nLANGUAGES\nGerman: mother tongue\nEnglish: fluent"], "expected": ["C2"], "shown": ["German, “mother tongue”"]}]""")   # per agent: its role, the texts asked, the answers expected, how a long text is shown

if not globals().get("CONNECTED") or "AGENTS" not in globals():
    print("Run the cells above first: Connect, then Set up your workspace.")
elif len(AGENTS) < len(AGENT_SPECS):
    _card("Some agents aren't in your workspace yet",
          [("ok" if s["role"] in AGENTS else "todo", s["title"], AGENTS[s["role"]]["name"] if s["role"] in AGENTS
            else "create it in AI Studio, then press Check again in step 2") for s in AGENT_SPECS])
else:
    _rows = []
    for _c in _CHECKS:
        try:
            for _text, _want, _shown, (_got, _label) in zip(_c["texts"], _c["expected"], _c["shown"], hrflow.tag(AGENTS[_c["role"]]["key"], _c["texts"])):
                _label = f"{_got} · {_label}" if _got else "Other"
                _ok = (_got or None) == (_want or None)
                _rows.append(("ok" if _ok else "todo", f"{_c['title']} · {_shown}",
                              _label if _ok else f"{_label} (expected {_want or 'Other'}: check the agent's taxonomy and description)"))
        except Exception as e:
            _rows.append(("err", f"{_c['title']} · {AGENTS[_c['role']]['name']}", str(e)))
    _ok = all(s == "ok" for s, _, _ in _rows)
    _card("Every agent answers" if _ok else "Something needs attention", _rows,
          "Setup done. Next: 4 · Live demo." if _ok else
          "An error on an agent: open it in AI Studio and press Deploy if its page shows it, then run this cell again."
          if any(s == "err" for s, _, _ in _rows) else
          "An answer in grey: check the agent's taxonomy, and that its description was pasted whole.")
```

## 💡 Good to know

- **One agent for every language.** The level agent reads the language named on its first line and the profile in its own language: no list of languages, no word list of a given language.
- **The normalizer always answers a code.** A name that is not one language (a level word, a nationality) still gets a code; the level step then usually finds no level, and the tag ends in `:other`.
- **What this notebook writes.** Step 2 creates the two taxonomies, on your click. The demo and the code only read profiles and call the agents: writing the tags to `profile["tags"]` is one call in the workflow that runs the pipeline.
